# Fairness without Imputation: A Decision Tree Approach for Fair Prediction with Missing Values

**Paper:** Jeong, Wang, & Calmon (AAAI-22)

This notebook implements the MIA (Missing Incorporated as Attribute) decision tree approach for fair machine learning with incomplete data.

## Key Concepts:
- **Problem:** Standard fair ML algorithms assume complete data; imputation before fairness interventions can introduce bias
- **Solution:** MIA decision tree that treats missing values as a category, with fairness-regularized objective
- **Benefits:** Avoid bias from imputation while maintaining fairness guarantees
